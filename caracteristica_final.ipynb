```python
import pandas as pd
import os
import sys
```

```python
df_caracteristicas = pd.read_csv("C:/Users/Santiago/OneDrive - Universidad EAFIT/Documents/EAFIT/CAOBA/PROYECTO_INTEGRADOR_2025II/DATA/PROCESSED/key_caracteristicas.csv", sep=";", encoding="latin1")
df_caracteristicas.head()
```

```python
df_hogar = pd.read_csv("C:/Users/Santiago/OneDrive - Universidad EAFIT/Documents/EAFIT/CAOBA/PROYECTO_INTEGRADOR_2025II/DATA/PROCESSED/key_hogar.csv", sep=";", encoding="latin1")
df_hogar.head()
```

```python
# Merge the dataframes while avoiding duplicate columns
df_merged = pd.merge(
    df_caracteristicas, 
    df_hogar, 
    on='primary_key', 
    how='left', 
    suffixes=('', '_y')
)

# Drop the duplicated/repeated columns from the second dataframe
columns_to_drop = [col for col in df_merged.columns if col.endswith('_y')]
df_merged = df_merged.drop(columns=columns_to_drop)

df_merged.head()
```
