```python
import pandas as pd
import statsmodels.api as sm
from imblearn.over_sampling import SMOTE
```

```python
df = pd.read_csv(r"C:\Users\ASUS\Downloads\df_para_modelo_II.csv")
```

```python
df
```

```python
df['trabajo_infantil'] = df['trabajo_infantil'].replace({False: 0, True: 1})
```

```python
df
```

```python
from sklearn.model_selection import train_test_split

# Dividir el conjunto de datos en entrenamiento y prueba
train, test = train_test_split(df, test_size=0.3, random_state=42)

# Dividir el conjunto de prueba en prueba y validación
test, validation = train_test_split(test, test_size=0.5, random_state=42)

# Verificar las dimensiones de cada conjunto
print(f"Entrenamiento: {train.shape}")
print(f"Prueba: {test.shape}")
print(f"Validación: {validation.shape}")
```

```python
validation_2 = validation.copy()
```

```python
# Eliminar las columnas 'key_ninos' y 'primary_key' de todas las particiones
train = train.drop(columns=['key_ninos', 'primary_key'])
test = test.drop(columns=['key_ninos', 'primary_key'])
validation = validation.drop(columns=['key_ninos', 'primary_key'])
```

```python
# Separar las características (X) y la variable objetivo (y) del conjunto de entrenamiento
X_train = train.drop(columns=['trabajo_infantil'])
y_train = train['trabajo_infantil']
```

```python
# Aplicar SMOTE para balancear las clases
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# Verificar la distribución de clases después del balanceo
print("Distribución de clases antes del balanceo:")
print(y_train.value_counts())
print("\nDistribución de clases después del balanceo:")
print(y_train_balanced.value_counts())
```

```python
# Agregar constante para el intercepto
X_train_balanced_with_const = sm.add_constant(X_train_balanced)

# Ajustar el modelo logístico con statsmodels
logit_model = sm.Logit(y_train_balanced, X_train_balanced_with_const).fit()

# Mostrar el resumen con coeficientes, errores estándar, z, p-valores, etc.
print(logit_model.summary())
```

```python
# Add a constant to the validation data for the intercept
X_validation_with_const = sm.add_constant(validation.drop(columns=['trabajo_infantil']))

# Make predictions using the logistic regression model
y_validation_pred_logit = logit_model.predict(X_validation_with_const)

# Display the predictions
print(y_validation_pred_logit)
```

```python
prediccion = y_validation_pred_logit.to_frame(name='prediccion')
```

```python
validation_2['prediccion'] = prediccion['prediccion'].values
```

```python
validation_2
```

```python
validation_2 = validation_2[['key_ninos', 'primary_key', 'prediccion']]
```

```python
df = pd.read_parquet(r'C:\Users\ASUS\Downloads\finallabeled_.parquet')
```

```python
# Realizar el merge entre validation_2 y df utilizando la columna key_ninos
merged_df = pd.merge(validation_2, df, on='key_ninos', how='inner')

# Mostrar las primeras filas del DataFrame resultante
print(merged_df.head())
```
