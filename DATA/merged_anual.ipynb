```python
import os
import pandas as pd
```

```python
# Merge Características Generales, Seguridad Social en Salud y Educación

# Ruta base
base_path = "C:/Users/Santiago/OneDrive - Universidad EAFIT/Documents/EAFIT/CAOBA/PROYECTO_INTEGRADOR_2025II/DATA/RAW"
folders = sorted(os.listdir(base_path))

df_list = []
sample_columns = None  # Para almacenar las columnas del primer archivo

for folder in folders:
    file_path = os.path.join(base_path, folder, "Características generales, seguridad social en salud y educación.CSV")
    if os.path.isfile(file_path):
        try:
            df = pd.read_csv(file_path, encoding="latin1", sep=";")  # usa sep=";" si aplica
            df["folder_id"] = folder  # opcional
            if sample_columns is None:
                sample_columns = df.columns.tolist()  # usar como plantilla
            df = df[[col for col in sample_columns if col in df.columns]]  # conservar columnas válidas
            df_list.append(df)
        except Exception as e:
            print(f"⚠️ Error leyendo {file_path}: {e}")

# Concatenar y reordenar columnas
merged_df = pd.concat(df_list, ignore_index=True)
merged_df = merged_df[sample_columns]

# Guardar CSV final
merged_df.to_csv("C:/Users/Santiago/OneDrive - Universidad EAFIT/Documents/EAFIT/CAOBA/PROYECTO_INTEGRADOR_2025II/DATA/PROCESSED/caracteristicas.csv", sep=";", index=False, encoding="latin1")
print(f"✅ Archivo guardado con {len(merged_df)} filas y columnas en orden estándar.")
```
